```python
from google.colab import files
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter

# Use the generated sample PDF directly
pdf_filename = "sample_policy.pdf"

# Load the PDF
loader = PyPDFLoader(pdf_filename)
raw_documents = loader.load()

print(f"Loaded {len(raw_documents)} pages from {pdf_filename}.")

# Text Splitting: Break documents into manageable chunks with overlap
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=200,
    add_start_index=True
)

docs = text_splitter.split_documents(raw_documents)
print(f"Split into {len(docs)} chunks for embedding.")
```

```python
!pip install -q langchain-huggingface

from langchain_chroma import Chroma
from langchain_huggingface import HuggingFaceEmbeddings

# Initialize embedding model
embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")

# Create a local vector database
vectorstore = Chroma.from_documents(
    documents=docs,
    embedding=embeddings,
    collection_name="professional_rag_db"
)

# Create a retriever from the vector store
retriever = vectorstore.as_retriever(
    search_type="similarity",
    search_kwargs={"k": 4} # Retrieve top 4 most relevant chunks
)

print("Vector database created and retriever initialized successfully!")
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnablePassthrough

# Initialize Gemini model
llm = ChatGoogleGenerativeAI(model="gemini-3.8-flash", temperature=0.3)

# Define a professional system prompt that reduces hallucinations
system_prompt = (
    "You are an expert assistant for question-answering tasks. "
    "Use the following pieces of retrieved context to answer "
    "the question. If you don't know the answer, say that you "
    "don't know. Use three sentences maximum and keep the "
    "answer concise.\n\n"
    "{context}"
)

prompt = ChatPromptTemplate.from_messages([
    ("system", system_prompt),
    ("human", "{input}"),
])

# Helper function to format retrieved documents into a single string
def format_docs(docs):
    return "\n\n".join(doc.page_content for doc in docs)

# Build the LCEL (LangChain Expression Language) chain
rag_chain = (
    {"context": retriever | format_docs, "input": RunnablePassthrough()}
    | prompt
    | llm
    | StrOutputParser()
)

print("RAG chain constructed successfully!")
```

```python
import time
from google.api_core.exceptions import GoogleAPIError

query = "What is the annual wellness stipend amount?"

# Retry mechanism to handle temporary 503 high-demand errors gracefully
max_retries = 3
delay = 5  # seconds to wait before retrying

for attempt in range(max_retries):
    try:
        response = rag_chain.invoke(query)
        print(response)
        break
    except GoogleAPIError as e:
        if "503" in str(e) and attempt < max_retries - 1:
            print(f"Gemini API is busy (503). Retrying in {delay} seconds (Attempt {attempt + 1}/{max_retries})...")
            time.sleep(delay)
            delay *= 2  # Exponential backoff
        else:
            print("Failed to query the model due to persistent API demand:")
            raise e
```

```python
from datasets import Dataset

# Define a small test set based on your sample policy document
eval_data = {
    "question": [
        "What is the annual wellness stipend amount?",
        "How many remote work days are allowed per week?"
    ],
    "ground_truth": [
        "The annual wellness stipend is $500 per year.",
        "Employees are allowed to work remotely up to 3 days per week."
    ]
}

dataset = Dataset.from_dict(eval_data)
print("Evaluation dataset created successfully!")
```

```python
answers = []
contexts = []

for query in dataset["question"]:
    # Get retrieved documents
    retrieved_docs = retriever.invoke(query)
    context_content = [doc.page_content for doc in retrieved_docs]

    # Generate response using your existing rag_chain
    response = rag_chain.invoke(query)

    answers.append(response)
    contexts.append(context_content)

# Bundle everything into a format Ragas expects
eval_dataset_dict = {
    "question": dataset["question"],
    "answer": answers,
    "contexts": contexts,
    "ground_truth": dataset["ground_truth"]
}

eval_results_dataset = Dataset.from_dict(eval_dataset_dict)
```

```python
import sys
import types

# Create a mock module to prevent ragas import crash
dummy_vertex = types.ModuleType("langchain_community.chat_models.vertexai")
dummy_vertex.ChatVertexAI = type("ChatVertexAI", (object,), {})
sys.modules["langchain_community.chat_models.vertexai"] = dummy_vertex

from ragas import evaluate
from ragas.metrics import faithfulness, answer_relevancy
from ragas.llms import LangchainLLMWrapper
from ragas.embeddings import LangchainEmbeddingsWrapper
from langchain_google_genai import ChatGoogleGenerativeAI, GoogleGenerativeAIEmbeddings
from ragas.run_config import RunConfig

# Setup evaluators
evaluator_llm = LangchainLLMWrapper(ChatGoogleGenerativeAI(model="gemini-3.8-flash", temperature=0))
evaluator_embeddings = LangchainEmbeddingsWrapper(GoogleGenerativeAIEmbeddings(model="models/text-embedding-004"))

# Run evaluation with only core metrics to prevent overloading limits
config = RunConfig(timeout=360, max_retries=20, max_wait=60)

result = evaluate(
    eval_results_dataset,
    metrics=[faithfulness, answer_relevancy],
    llm=evaluator_llm,
    embeddings=evaluator_embeddings,
    run_config=config,
)

# Display your populated scoring table
df = result.to_pandas()
display(df)
```
